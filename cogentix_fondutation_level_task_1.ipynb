```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("/content/sales_data_sample.csv", encoding="latin1")

print("=== Sales Summary Statistics ===")
print("Total Sales (Sum):", df["SALES"].sum())
print("Average Sales (Mean):", df["SALES"].mean())
print("Maximum Sale Value:", df["SALES"].max())
print("Minimum Sale Value:", df["SALES"].min())

top_products = df.groupby("PRODUCTLINE")["SALES"].sum().sort_values(ascending=False)
print("\n=== Top Performing Product Lines ===")
print(top_products)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

top_products.plot(kind="bar", ax=axes[0], color="skyblue", edgecolor="black")
axes[0].set_title("Sales by Product Line")
axes[0].set_ylabel("Total Sales ($)")
axes[0].tick_params(axis="x", rotation=45)

region_sales = df.groupby("COUNTRY")["SALES"].sum().head(6)
axes[1].pie(region_sales, labels=region_sales.index, autopct="%1.1f%%", startangle=140)
axes[1].set_title("Top 6 Countries Sales Share")

plt.tight_layout()
plt.show()
```
